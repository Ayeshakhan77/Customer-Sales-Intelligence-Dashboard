```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df.columns
```

```python
#Feature engineering

#create a new column age_group
labels = ['young', 'adult', 'mid_aged', 'senior']

bins = [0, 20, 35, 55, 100]
df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels, right=False)
```

```python
df[['age', 'age_group']].head(20)
```

```python
df['frequency_of_purchases'].unique()
```

```python
#create column purchase frequency

# Mapping dictionary (frequency → number per year)
freq_map = {
    'Weekly': 7,
    'Bi-Weekly': 14,        # every 2 weeks
    'Fortnightly': 14,      # same as bi-weekly
    'Monthly': 30,
    'Quarterly': 90,
    'Every 3 Months': 90,    # same as quarterly
    'Annually': 365
}

# Create new numeric column
df['purchase_freq'] = df['frequency_of_purchases'].map(freq_map)
```

```python
df[['purchase_freq', 'frequency_of_purchases']].head(20)
```

```python
df[['discount_applied', 'promo_code_used']].head(20)
```

```python
# Checking wether both columns are equal, if yes then we dont need both we will keep just 'discount_applied' column

print((df['discount_applied'] == df['promo_code_used']).all())

df = df.drop('promo_code_used', axis=1)
```
